### EDA on Hotel bookings 


```python
#impoting pandas 
import pandas as pd
```

```python
#read dataset 
hotel_bookings = pd.read_csv('hotel_bookings.csv')
```

```python
hotel_bookings
```

```python
hotel_bookings.columns
```

```python
#remove is_canceled 
hotel_bookings.drop(columns=['is_canceled'],inplace=True)
```

```python
hotel_bookings
```

```python
hotel_bookings.rename({'adults':'num_adults','children':'num_children','babies':'num_babies'},axis=1,inplace=True)
hotel_bookings.head(10)
```

```python
#finding the percentage of nan {no values}
hotel_bookings.isnull().sum() *100  / len(hotel_bookings)
```

```python
#we drop country becsuse there is many NAN values
hotel_bookings.drop(columns=['country'],inplace=True)
hotel_bookings['num_children'].dropna(inplace=True)
```

```python
hotel_bookings.isnull().sum() *100  / len(hotel_bookings)
```

```python
hotel_bookings['agent']=hotel_bookings['agent'].fillna(-1)
hotel_bookings['company']=hotel_bookings['company'].fillna('unknown')
hotel_bookings['num_children']=hotel_bookings['num_children'].fillna(1)
```

```python
hotel_bookings['company']
hotel_bookings.isnull().sum() *100  / len(hotel_bookings)
```

```python
hotel_bookings.dtypes
```

```python
#we change it because it has wrong datatype
hotel_bookings = hotel_bookings.astype({'num_children':'int'})
```

hotel_bookings.dtypes

```python
#bin columns ranged as ranges
hotel_bookings['lead_time'].unique()
hotel_bookings['lead_time'].describe()
bins = [0,100,200,300,400,500,600,700,800]
labels = ['0-100','101-200','201-300','301-400','401-500','501-600','601-700','701-800']
hotel_bookings['lead_time_binned'] = pd.cut(hotel_bookings['lead_time'],bins=bins,labels=labels)
hotel_bookings[['lead_time','lead_time_binned']]
```

```python
#we checked is there any difference or other values like symbols or anything
hotel_bookings['hotel'].unique
```

```python
#checking and removing duplicated rows
hotel_bookings.loc[hotel_bookings.duplicated(keep=False)]
hotel_bookings.drop_duplicates(keep='first',inplace=True)
```
