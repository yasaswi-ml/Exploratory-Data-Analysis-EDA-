## 3. Central Tendency & Dispersion

Statistical measures help us understand the distribution and spread of numerical data.

### Central Tendency
- Mean: Average of all observations.
- Median: Middle value of sorted observations.
- Mode: Most frequently occurring value.

### Dispersion
- Range: Difference between maximum and minimum values.
- IQR: Difference between the third and first quartiles.
- Variance: Average squared deviation from the mean.
- Standard Deviation: Measures the spread of values around the mean.
- Mean Absolute Deviation: Average absolute deviation from the mean.

```python
# Replace with your numerical column
target_col = "numeric_column"

# Central Tendency
mean_val = data[target_col].mean()
median_val = data[target_col].median()
mode_val = data[target_col].mode()[0]

# Dispersion
data_range = data[target_col].max() - data[target_col].min()

Q1 = data[target_col].quantile(0.25)
Q3 = data[target_col].quantile(0.75)

IQR = Q3 - Q1

variance = data[target_col].var()
std_dev = data[target_col].std()

mad = (
    data[target_col] - data[target_col].mean()
).abs().mean()

print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)
print("Range:", data_range)
print("IQR:", IQR)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
print("Mean Absolute Deviation:", mad)
```

## 4. Categorical Data Analysis

Categorical analysis helps us understand the frequency and distribution of different categories.

### Important Operations
- value_counts(): Counts occurrences of each category.
- normalize=True: Calculates category proportions.
- nunique(): Counts the number of unique categories.

```python
# Replace with your categorical column
cat_col = "categorical_column"

# Frequency Table
counts = data[cat_col].value_counts()

print("--- CATEGORY COUNTS ---")
print(counts)

# Category Proportions
props = data[cat_col].value_counts(normalize=True)

print("\n--- CATEGORY PROPORTIONS ---")
print(props)

# Percentage Distribution
percentages = data[cat_col].value_counts(normalize=True) * 100

print("\n--- CATEGORY PERCENTAGES ---")
print(percentages)
```

### 5.6 Correlation Heatmap

A correlation heatmap visualizes the correlation between numerical variables.

Correlation values range from -1 to +1.

- +1: Perfect positive correlation.
- -1: Perfect negative correlation.
- 0: No linear correlation.

```python
correlation = data.select_dtypes(
    include=np.number
).corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
plt.close()
```

## 6. Missing Values Analysis

Missing values occur when observations are unavailable for certain features.

Identifying missing values is an important step before preprocessing.

Important operations:
- isnull(): Identifies missing values.
- sum(): Counts missing values.
- mean(): Calculates missing value proportions.

```python
# Missing values count
missing_count = data.isnull().sum()

print("--- MISSING VALUES COUNT ---")
print(missing_count)

# Missing values percentage
missing_percent = (
    data.isnull().mean() * 100
).sort_values(ascending=False)

print("\n--- MISSING VALUES PERCENTAGE ---")
print(missing_percent)

# Rows containing missing values
missing_rows = data[data.isnull().any(axis=1)]

print("\nTotal rows with missing values:", len(missing_rows))
```

## 7. Duplicate Values Analysis

Duplicate records are identical rows appearing more than once in a dataset.

They can affect statistical analysis and model training.

Important operations:
- duplicated(): Identifies duplicate rows.
- drop_duplicates(): Removes duplicate rows.

```python
# Count duplicate rows
duplicate_count = data.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)

# Remove duplicates
data = data.drop_duplicates()

print("Shape After Removing Duplicates:", data.shape)
```

## 8. Outlier Detection

Outliers are observations that significantly differ from the majority of data points.

The Interquartile Range (IQR) method is commonly used for outlier detection.

### Formula

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 * IQR

Upper Bound = Q3 + 1.5 * IQR

Values outside these boundaries are considered potential outliers.

```python
Q1 = data[target_col].quantile(0.25)
Q3 = data[target_col].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data[target_col] < lower_bound) |
    (data[target_col] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

## 9. Data Cleaning

Data cleaning involves correcting inconsistencies and handling problematic observations.

Common operations:
- Replacing incorrect values.
- Removing unwanted rows.
- Handling missing values.
- Removing duplicate records.
- Handling outliers.

Note: Cleaning decisions should depend on the dataset and domain knowledge.

```python
# Replace specific values
 data["column_name"] = data["column_name"].replace("o", 0)

# Remove rows based on a condition
 data = data[data["column_name"] > 0]

# Replace negative values with zero
 data.loc[data["column_name"] < 0, "column_name"] = 0

# Check missing values
 data = data.dropna()

# Check duplicates
 data = data.drop_duplicates()
```
